```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img_bl = cv2.imread('./assets/PCD_UTS_PHOTO.jpg')
img_gray = cv2.cvtColor(img_bl, cv2.COLOR_BGR2GRAY)
```

```python
alpha_cerah = 1.0
beta_cerah = 60   
img_cerah = cv2.convertScaleAbs(img_gray, alpha=alpha_cerah, beta=beta_cerah)
```

```python
alpha_kontras = 1.8 
beta_kontras = 0    
img_kontras = cv2.convertScaleAbs(img_gray, alpha=alpha_kontras, beta=beta_kontras)
```

```python
alpha_ck = 1.5
beta_ck = 40
img_ck = cv2.convertScaleAbs(img_gray, alpha=alpha_ck, beta=beta_ck)
```

```python
plt.figure(figsize=(15, 10))
plt.subplot(231), plt.imshow(cv2.cvtColor(img_bl, cv2.COLOR_BGR2RGB)), plt.title('Gambar Asli')
plt.subplot(232), plt.imshow(img_gray, cmap='gray', vmin=0, vmax=255), plt.title('Gambar Gray')
plt.subplot(233), plt.imshow(img_cerah, cmap='gray', vmin=0, vmax=255), plt.title('Gambar Gray yang Dipercerah')
plt.subplot(234), plt.imshow(img_kontras, cmap='gray', vmin=0, vmax=255), plt.title('Gambar Gray yang Diperkontras')
plt.subplot(235), plt.imshow(img_ck, cmap='gray', vmin=0, vmax=255), plt.title('Gambar Gray yang Dipercerah & Diperkontras')
plt.tight_layout()
plt.show()
```
